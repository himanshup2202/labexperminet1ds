QUESTION 21

```python
sales = np.array(
    [[120, 150, 180], [140, 160, 175], [135, 155, 190], [150, 170, 200]]
)

daily_sales = sales.sum(axis=1)
product_sales = sales.sum(axis=0)
highest_day = np.argmax(daily_sales) + 1  #

print("Daily Sales (Days 1-4):", daily_sales)
print("Product Sales (Products 1-3):", product_sales)
print(
    f"Highest Combined Sales: Day {highest_day} ({daily_sales[highest_day-1]} units)"
)
```

QUESTION 23

```python
weights = np.array([48, 62, 75, 81, 54, 69, 90])

above_70 = weights[weights > 70]
replaced_weights = np.where(weights < 55, 55, weights)

print("Weights > 70:", above_70)
print("Modified Array:", replaced_weights)
print("Original Array Unchanged:", weights)
```

QUESTION 24

```python
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])

total_cost = quantity * price
overall_bill = total_cost.sum()
pct_contrib = (total_cost / overall_bill) * 100

print("Total Cost per Product:", total_cost)
print("Overall Bill:", overall_bill)
print("Percentage Contribution (%):", np.round(pct_contrib, 2))
```
